```python
###reading file

def read_fil(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'a.txt'

for Line in read_fil(file_path):
    print(Line.strip())
```

```python
def my_decorator(func):
    def wrapper():
        print("somrthing")
        func()
        print("hello")
    return wrapper()
    
```

```python
@my_decorator
def say_hello():
    print("hello1")
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*agrs,**kwargs):
            for _ in range(n):
                func(*agrs,**kwargs)
            return wrapper
        return decorator    
```

```python
@repeat(3)
def say_hello():
    print("hello")
```
