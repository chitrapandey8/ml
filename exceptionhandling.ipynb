```python
#try, except, else block
try:
    num= int(input("enter the nnumber"))
    result=10/num
except ValueError:
    print("this is not a valid number")
except ZeroDivisionError:
    print("zero division error") 
except Exception as ex:
    print(ex)
else:
    print(f"the result is {result}")  ##whenever error happens else doesnot excute it got handled by any of the except block
    
```

```python
##try, except, else, finally
try:
    num= int(input("enter the nnumber"))
    result=10/num
except ValueError:
    print("this is not a valid number")
except ZeroDivisionError:
    print("zero division error") 
except Exception as ex:
    print(ex)
else:
    print(f"the result is {result}") 
finally:
    print("this has to excute")
    print("excution completed")
```

```python
#file handling and exception ahndling
try:
    file=open('example11.txt','r')
    content = file.read()
    print(content)
except FileNotFoundError:
    print("the file does not exit")
except Exception as ex:
    print(ex)
finally:
    if 'file' in locals() and not file.closed():
        file.close()     #making sure we  close the file
        print('file close')
    
```
