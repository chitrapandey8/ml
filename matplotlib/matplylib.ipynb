```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

plt.plot(x,y)
plt.xlabel('x axis') ##labls deta hai x 0r y axis ko
plt.ylabel('y axis')
plt.title("basic line")
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

##cresting customised line plot

plt.plot(x,y, color= 'red', linestyle = 'solid',marker='o',linewidth = '2')
plt.grid(True)
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
y2= [1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.plot(x,y,y2, color = "green")
plt.xlabel("X LABEL")

```

```python

x=[1,2,3,4,5]
y=[1,4,9,16,25]
y2= [1,2,3,4,5]


plt.Subplot(1,2,2)
plt.plot(x,y2, color = "red")
```

```python
#barplot
categories = ['A','B','C','D','E']
values  = [5,7,3,8,6]

plt.bar(categories, values,color = "red")
```

```python
data = [1,2,3,5,3,4,5,5,3,4,5,3,4,5,4]
plt.hist(data,bins=5,edgecolor = "black")
```

```python
##sample data

x = [1,2,3,4,5]
y = [2,3,4,5,5]
plt.scatter(x,y)
```

```python
#pychart

labels = ['A','B','C','D','E']
size  = [10,20,40,10,15]
colors=["red","blue","black","yellow","pink"]
explode = (0,2,0,0,0)
plt.pie(size,labels=labels,colors=colors,explode=explode,shadow = 'true')

plt.show()
```
